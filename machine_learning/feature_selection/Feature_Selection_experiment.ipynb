```python
import pandas as pd
import numpy as np
from sklearn.feature_selection import SelectKBest
from sklearn.feature_selection import f_classif , chi2
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import load_iris
```

```python
data = load_iris()
```

```python
X = data.data
y=data.target
```

```python
df = pd.DataFrame(X, columns=data.feature_names)
```

```python
df['target'] = data.target_names[y]
```

```python
scaler = StandardScaler()
x_train , x_test , y_train, y_test = train_test_split(X, y , random_state=42, test_size=0.2)
```

```python
x_train_scaler=scaler.fit_transform(x_train)
x_test_scaler=scaler.transform(x_test)
```

```python
# SelectKBest method apply
x_new_features = SelectKBest(score_func=f_classif, k=3) # f_classif score
x_new_features.fit(x_train_scaler, y_train)
```

```python
selectK_x_fea=x_new_features.fit_transform(x_train_scaler, y_train)
selectK_x_fea_test=x_new_features.fit_transform(x_test_scaler, y_test)
```

```python
x_new_features.get_support(indices=True)
```

```python
# use Wrapper method

from sklearn.feature_selection import RFE
from sklearn.linear_model import LogisticRegression

# intilize the logistic regressio model for estimator
estimator=LogisticRegression(max_iter=1000)

rfe=RFE(estimator , step=1)
rfe.fit(x_train_scaler, y_train)

```

```python
rfe_x_train=rfe.fit_transform(x_train_scaler, y_train)
rfe_x_test=rfe.transform(x_test_scaler)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import  accuracy_score, classification_report

RFC=RandomForestClassifier(n_estimators=100 , max_depth=2)
```

```python
RFC.fit(selectK_x_fea, y_train)
```

```python
y_test_pre=RFC.predict(selectK_x_fea_test)
```

```python
accuracy_score(y_test, y_test_pre)
print(classification_report(y_test_pre, y_test))
```

```python
# now train model RFE
RFC.fit(rfe_x_train, y_train)
```

```python
rfe_y_pre=RFC.predict(rfe_x_test)
```

```python
print(classification_report(rfe_y_pre, y_test))
```
