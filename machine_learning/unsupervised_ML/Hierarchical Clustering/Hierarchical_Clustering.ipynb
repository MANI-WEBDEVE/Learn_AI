```python
import numpy as np
from sklearn.cluster import AgglomerativeClustering
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage

X = np.array([
    [1, 1], [1, 2],      # Group 1 (left-bottom)
    [4, 4], [5, 5],      # Group 2 (middle)
    [8, 8], [9, 9],      # Group 3 (right-top)
    [1, 8], [2, 9]       # Group 4 (left-top)
])

# Model banana: 4 clusters chahiye, average linkage use
model = AgglomerativeClustering(
    n_clusters=4,
    linkage='average',   # 'single', 'complete', 'ward' 
    metric='euclidean'
)

# Predict karo har point ka cluster number
labels = model.fit_predict(X)

print("Har point ka cluster number:")
for i, label in enumerate(labels):
    print(f"Point {i}: Cluster {label}")

# Output example: [0 0 1 1 2 2 3 3] → 4 clear groups!
```

```python
Z = linkage(X, method='average', metric='euclidean')

plt.figure(figsize=(10, 5))
dendrogram(Z, labels=['A','B','C','D','E','F','G','H'])
plt.title('Hierarchical Clustering ka Family Tree')
plt.xlabel('Points')
plt.ylabel('Distance (Kitni doori pe merge hua)')
plt.axhline(y=3, color='red', linestyle='--', label='Yahan kaatoge to 4 clusters mileinge')
plt.legend()
plt.show()
```
