### DBSCAN (Density Based Spetial Cluster Application with Noise).



```python
# import the important libraries

# DBSCAN its avaliable in sklearn.cluster module
# make_moons data set avaliable in sklearn.datasets module

import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons
```

```python

# 1. Data Create karte hain (Do chandrama shapes)
# K-Means yahan fail ho jata hai, par DBSCAN aaram se pakad lega.
X, y = make_moons(n_samples=500, noise=0.05, random_state=42)
```

```python
X.shape
```

```python
plt.scatter(X[:,0], X[:, 1])
```

```python

# 2. DBSCAN Model Setup karte hain
# eps=0.3: Points ke beech max distance
# min_samples=5: Cluster banne ke liye kam se kam 5 points chahiye
dbscan = DBSCAN(eps=0.3, min_samples=5)

# 3. Model ko fit karte hain
dbscan.fit(X)
```

```python

# 4. Results nikalte hain
labels = dbscan.labels_       # Har point ko cluster number mila (-1 means Noise)
core_samples = dbscan.core_sample_indices_ # Core points ki list

# 5. Plotting (Visualize karne ke liye)
plt.figure(figsize=(8, 6))

# Unique labels nikalo (Noise -1 ko alag color denge)
unique_labels = set(labels)
colors = plt.cm.Spectral(np.linspace(0, 1, len(unique_labels)))

for k, col in zip(unique_labels, colors):
    if k == -1:
        # Black color for Noise
        print('k-1')
        col = [0, 0, 0, 1]

    # Us label ke saare points ko dhoondo
    print(k)
    class_member_mask = (labels == k)

    # Plot karo
    plt.plot(X[class_member_mask, 0], X[class_member_mask, 1],
             'o', markerfacecolor=tuple(col), markeredgecolor='k', markersize=6)

plt.title(f'DBSCAN Clustering (Total Clusters: {len(unique_labels) - 1})')
plt.show()
```
